# EEG-Flow — 6. ICA selection comparison and consensus

Loads the ICA selections from both reviewers and plots components where
they disagreed. Both reviewers review disagreements together and reach a
consensus, saved as the final step6 ICA files used in step 7.

**Note (v2):** for participants already preprocessed in v1, this step is
not run. Consensus selections are copied directly from v1 `derivatives/`
by `convert_v1_to_v2_datasetfix.ipynb`. Agreement plots (`plots/ica/`)
are also copied.

Outputs per run:
- `_step6_reviewed_1st_ica.fif` — consensus 1st ICA (mastoid cleaning)
- `_step6_reviewed_2nd_ica.fif` — consensus 2nd ICA (EEG cleaning)
- Reviewer comparison plots in `plots/ica/`
